```python
CREATE TABLE artists (
    id		INTEGER PRIMARY KEY AUTOINCREMENT,
    name	TEXT NOT NULL UNIQUE

);

CREATE TABLE albums (
    id			INTEGER PRIMARY KEY AUTOINCREMENT,
    artist_id	INTEGER NOT NULL,
    title		TEXT NOT NULL,
    release_year	INTEGER NOT NULL,
    FOREIGN KEY (artist_id)
        REFERENCES artists(id)
        ON DELETE CASCADE
        ON UPDATE CASCADE,
    CHECK (release_year BETWEEN 1900 AND 2100),
    UNIQUE (artist_id, title, release_year)

);

CREATE TABLE tracks (
    id				INTEGER PRIMARY KEY AUTOINCREMENT,
    album_id		INTEGER NOT NULL,
    title			TEXT NOT NULL,
    duration_seconds	INTEGER NOT NULL,
    FOREIGN KEY (album_id)
        REFERENCES albums(id)
        ON DELETE CASCADE
        ON UPDATE CASCADE,
    CHECK (duration_seconds > 0),
    UNIQUE (album_id, title)
);
```

```python
%%sql
select c.last_name, ci.city from customer c
JOIN address a ON c.address_id = a.address_id
JOIN city ci ON a.city_id = ci.city_id
WHERE ci.city = "London"
```

```python
%%sql
SELECT f.title, c.name FROM film f
JOIN film_category fc ON f.film_id = fc.film_id
JOIN category c ON fc.category_id = c.category_id
```

```python
%%sql
SELECT f.title, a.actor_id FROM film f
JOIN film_actor fa ON f.film_id = fa.film_id
JOIN actor a ON a.actor_id = fa.actor_id
WHERE a.actor_id = 5
```

```python
%%sql
SELECT f.film_id, f.title, count(r.rental_id) how_much FROM film f
JOIN inventory i ON f.film_id = i.film_id
JOIN rental r ON i.inventory_id = r.inventory_id
GROUP BY f.film_id
ORDER BY how_much DESC
```
